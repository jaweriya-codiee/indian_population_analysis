```python
female_total = data['Female'].sum()
print("total no.of female :",female_total)
```

```python
male_total = data["Male"].sum()
print("total no. of males :",male_total)
```

```python
gender_data = pd.DataFrame({"gender": ["Female","Male"],"population":[female_total,male_total]})
```

```python
gender_data
```

```python
plt.figure(figsize=(5,4))
plt.bar(gender_data["gender"],gender_data["population"])
plt.xlabel("gender"),plt.ylabel("population")
plt.title("Population of india")
plt.show()
```
